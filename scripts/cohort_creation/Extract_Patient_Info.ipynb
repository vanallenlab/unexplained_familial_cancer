```python
person_id = "XX"
data_version = "C2022Q4R13"

query = f"""
SELECT
    co.*,
    c.concept_name AS condition_concept_name,
    c.concept_code AS condition_concept_code,
    c.vocabulary_id AS condition_vocabulary_id,
    cs.concept_name AS source_concept_name,
    cs.concept_code AS source_concept_code,
    cs.vocabulary_id AS source_vocabulary_id
FROM `wb-silky-artichoke-2408.{data_version}.condition_occurrence` co
LEFT JOIN `wb-silky-artichoke-2408.{data_version}.concept` c
    ON co.condition_concept_id = c.concept_id
LEFT JOIN `wb-silky-artichoke-2408.{data_version}.concept` cs
    ON co.condition_source_concept_id = cs.concept_id
WHERE CAST(co.person_id AS STRING) = '{person_id}'
ORDER BY co.condition_start_date, co.condition_concept_id
"""

person_records = pd.read_gbq(query)

print(f"Total condition records: {len(person_records):,}")

pd.set_option("display.max_columns", None)
pd.set_option("display.max_rows", None)
pd.set_option("display.max_colwidth", None)

display(person_records[person_records["condition_concept_name"].str.contains("malignant neoplas", case=False, na=False)][["person_id", "condition_concept_name"]])
```
